```python
def grade(score):

    if score>=90:
        return "A"
    elif score>=80:
        return "B"
    elif score>=70:
        return "C"
    elif score>=60:
        return "D"
    else:
        return "F"

print(grade(87))
```

```python
def average(scores):
    return sum(scores)/len(scores)

def grade(avg):

    if avg>=90:
        return "A"
    elif avg>=80:
        return "B"
    elif avg>=70:
        return "C"
    elif avg>=60:
        return "D"
    else:
        return "F"

name=input("이름 : ")

scores=[]

for i in range(3):
    score=int(input(f"{i+1}번째 점수 : "))
    scores.append(score)

avg=average(scores)

print("="*35)
print("학생 성적")
print("="*35)
print("이름 :",name)
print("평균 :",round(avg,2))
print("학점 :",grade(avg))
print("="*35)
```
