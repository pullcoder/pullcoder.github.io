```python
# 과제 3
# 원의 넓이
# 원의 넓이를 계산하는 함수를 작성하세요. 공식: 넓이 = π × r² (π는 3.141592 사용)

def circle_area(radius):
    return round(3.1415926535 * radius * radius,2)
circle_area(int(input("원의 넓이: ")))
```

```python
# 과제 4
# 세 과목 평균
# 세 과목의 점수를 입력받아 평균을 계산하는 함수를 작성하세요.

def average(kor, eng, python):
    return (kor + eng + python) / 3

print(average(int(input("한국어 점수: ")), int(input("영어 점수: ")), int(input("python 점수: "))))
```

```python
# 과제 6
# 최대값 / 최소값 함수
# 리스트를 입력받아 최대값과 최소값을 반환하는 함수를 작성하세요.

# max_min(score)  ·  score=[90,75,88,100]

def max_min(score):
    return max(score), min(score)

lst = []
print("0 이면 종료")
while True:
    num = int(input("리스트에 담을 수: "))
    if num == 0:
        break
    lst.append(num)
a,b = max_min(lst)
print(f"최대값 : {a} / 최소값 : {b}")
```

```python
# 과제 7
# 학점 반환 함수
# 학생 이름과 점수를 입력받아 학점을 반환하는 함수를 작성하세요.

def student_score_return(student_name, scores):
    grade_point = 0

    for score in scores:
        if score >= 95:
            grade_point += 4.5
        elif score >= 90:
            grade_point += 4.0
        elif score >= 85:
            grade_point += 3.5
        elif score >= 80:
            grade_point += 3.0
        elif score >= 75:
            grade_point += 2.5
        elif score >= 70:
            grade_point += 2.0
        elif score >= 65:
            grade_point += 1.5
        elif score >= 60:
            grade_point += 1.0
        elif score >= 55:
            grade_point += 0.5
        elif score >= 50:
            grade_point += 0.0

    return round(grade_point / 4, 2)


student_name = input("이름: ")
scores = []

for subject_number in range(1, 5):
    subject_score = int(input(f"{subject_number}번째 점수: "))
    scores.append(subject_score)

print(student_score_return(student_name, scores))
```

```python
# 과제 8
# 구구단 함수
# 구구단을 출력하는 함수를 작성하세요.

def gugudan(gugu):
    print(f" {gugu}단 출력 ")
    for i in range(1,10):
        print(f"{gugu} * {i} = {gugu * i}")

gugudan(int(input("출력할 구구단: ")))
```

```python
# 과제 9
# 계산기 함수
# 사칙연산을 수행하는 계산기 함수를 작성하세요.

def calculator(a, b, op):
    if op == "+":
        return a + b
    elif op == "-":
        return a - b
    elif op == "*":
        return a * b
    elif op == "/":
        return a / b
    else:
        return "연산자가 잘못되었습니다."
    
print(calculator(int(input("숫자: ")), int(input("숫자: ")), input("연산자: ")))
```
