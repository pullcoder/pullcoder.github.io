```python
import os

os.makedirs("output", exist_ok=True)

# 5. 파일 쓰기 (write)
with open("output/hello.txt", "w", encoding="utf-8") as f:
    f.write("안녕하세요.")
    f.write("나는 여주대학교 AI 소프트웨어학과에 다니는 홍길동입니다.")
# 6. 여러 줄 저장
with open("output/student.txt", "w", encoding="utf-8") as fn:
    fn.write("홍길동\n")
    fn.write("김철수\n")
    fn.write("이영희\n")

# 7. 파일 전체 읽기 (read)
with open("output/hello.txt", "r", encoding="utf-8") as f:
    text = f.read()
with open("output/student.txt", "r", encoding="utf-8") as fn:
    textn =fn.read()
    print(text)
    print(textn)

# 8. 한 줄씩 읽기 (readline)
with open("output/student.txt", "r", encoding="utf-8") as f:
    print(f.readline(), end="")
    print(f.readline(), end="")
    print(f.readline(), end="")

# 9. 모든 줄을 리스트로 읽기 (readlines)
with open("output/student.txt", "r", encoding="utf-8") as f:
    lines = f.readlines()
    print(lines)

# 10. 반복문으로 읽기
# for 반복문은 파일에서 한 줄씩 읽어 line 변수에 저장합니다.
# strip()은 줄 끝의 줄바꿈 문자와 양쪽 공백을 제거한 뒤 해당 줄을 출력합니다.
with open("output/student.txt", "r", encoding="utf-8") as f:
    for line in f:
        print(line.strip())

# 11. 내용 추가하기 ("a" 모드)
with open("output/student.txt", "a", encoding="utf-8") as f:
    f.write("박민수\n")

# 12. with 문으로 파일 자동 닫기
with open("output/student.txt", "r", encoding="utf-8") as f:
    print(f.read())

# 13. UTF-8 인코딩 처리
with open("output/student.txt", "w", encoding="utf-8") as f:
    f.write("안녕하세요.\n")
with open("output/student.txt", "r", encoding="utf-8") as f:
    print(f.read())

# 14. 사용자 입력값 저장
name = input("이름: ")
score = input("점수: ")
with open("output/student.txt", "a", encoding="utf-8") as f:
    f.write(f"{name},{score}\n")

# 15. CSV(Comma-Separated Values) 형태 저장
with open("output/score.csv", "w", encoding="utf-8") as f:
    f.write("이름,Python,AI\n")
    f.write("홍길동,95,90\n")
    f.write("김철수,88,85\n")

# 16. 파일 복사
with open("output/student.txt", "r", encoding="utf-8") as source:
    with open("output/copy.txt", "w", encoding="utf-8") as target:
        target.write(source.read())

# 17. 학생 성적 데이터 저장 및 읽기
students = ["홍길동,95", "김철수,88", "이영희,100"]
with open("output/student.txt", "w", encoding="utf-8") as f:
    for student in students:
        f.write(student + "\n")
with open("output/student.txt", "r", encoding="utf-8") as f:
    for line in f:
        name, score = line.strip().split(",")
        print(name, score)

# 18. 파일 읽기 예외 처리
try:
    with open("output/test.txt", "r", encoding="utf-8") as f:
        print(f.read())
except FileNotFoundError:
    print("파일이 없습니다.")
```
