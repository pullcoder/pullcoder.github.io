```python
# 과제 5
# 0 입력 시 종료 — 누적 합계
# 사용자가 0을 입력할 때까지 숫자를 계속 입력받고, 입력한 숫자의 합을 출력하세요.

# 숫자 : 10
# 숫자 : 5
# 숫자 : 8
# 숫자 : 0

# 합계 : 23

k = 0
print("누적합계 / 0 종료")
while True:
    num = int(input("더할 값: "))
    if num == 0:
        break
    k += num
print(k)
```

```python
# 과제 6
# 90점 이상 학생만 출력
# 학생 이름과 점수를 딕셔너리로 저장한 후, 90점 이상인 학생만 출력하세요.

students = {
    "홍길동":95,
    "김철수":88,
    "이영희":100,
    "박민수":75
}

for i in students:
    if students[i] >= 90:
        print(i, students[i])
```

```python
# 과제 7
# for, while ATM
# 다음 메뉴를 완성하고 종료시까지 while, for를 이용하여 반복하기

# 초기 잔액
balance = 100000

# 거래 내역을 저장할 리스트
transactions = []

while True:
    print("\n===== ATM =====")
    print("1. 잔액 확인")
    print("2. 입금")
    print("3. 출금")
    print("4. 종료")
    print("==============")

    choice = input("메뉴를 선택하세요: ")

    # 1. 잔액 확인
    if choice == "1":
        print(f"현재 잔액: {balance:,}원")

    # 2. 입금
    elif choice == "2":
        money = int(input("입금할 금액: "))
        balance += money
        transactions.append("입금")
        print(f"{money:,}원이 입금되었습니다.")
        print(f"현재 잔액: {balance:,}원")

    # 3. 출금
    elif choice == "3":
        money = int(input("출금할 금액: "))

        # 출금 가능 여부 판단
        if money <= balance:
            balance -= money
            transactions.append("출금")
            print(f"{money:,}원이 출금되었습니다.")
            print(f"현재 잔액: {balance:,}원")
        else:
            print("잔액이 부족합니다.")

    # 4. 종료
    elif choice == "4":
        print("ATM을 종료합니다.")
        break

    # 잘못된 메뉴
    else:
        print("잘못된 메뉴입니다.")


# 거래 횟수 출력
print("\n===== 거래 내역 =====")

count = 0

for transaction in transactions:
    count += 1
    print(f"{count}. {transaction}")

print(f"총 거래 횟수: {count}회")
```
