```python
# 과제 1
# 기초 — 성인 판별
# 사용자의 나이를 입력받아 다음과 같이 출력하세요.

# 19세 이상 → 성인
# 19세 미만 → 미성년자

age = int(input("나이입력"))
if age>19:
    print("성인")
else:
    print("미성년자")
```

```python
# 과제 2
# 짝수 / 홀수
# 정수를 입력받아 짝수인지 홀수인지 출력하세요.

num = int(input("정수입력"))
if num % 2 != 0:
    print("홀")
else:
    print("짝")
```

```python
# 과제 3
# 성적 등급
# 점수를 입력받아 학점을 출력하세요.

# 점수	학점
# 90~100	A
# 80~89	B
# 70~79	C
# 60~69	D
# 0~59	F

score = int(input("점수입력"))

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
elif score >= 60:
    grade = "D"
else:
    grade = "F"
print(grade)
```

```python
# 과제 4
# 윤년 판별
# 연도를 입력받아 윤년인지 판별하세요.

# 💡 힌트: 윤년은 4의 배수이면서 100의 배수가 아니거나, 400의 배수입니다.

year = int(input("연도를 입력하세요."))
if year % 4 == 0:
    if year % 400 == 0:
        print("o")
    elif year % 100 != 0:
        print("o")
    else:
        print("x")
else:
    print("x")
```

```python
# 과제 5
# 간이 계산기
# 두 수와 연산자(+, -, *, /)를 입력받아 계산 결과를 출력하세요.

# 첫 번째 수 : 20
# 두 번째 수 : 5
# 연산자 : *

# 결과 : 100

num1 = int(input("숫자 1"))
num2 = int(input("숫자 2"))
operator = input("연산자")

if operator == "+":
    print(num1+num2)
elif operator == "-":
    print(num1-num2)
elif operator == "*":
    print(num1*num2)
elif operator == "/":
    if num2 == 0:
        print("0으로 나눌 수 없음.")
    else:
        print(num1/num2)
```
