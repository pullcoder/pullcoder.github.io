```python
# 과제 1번.
# hello.txt 파일을 생성하고 다음 두 줄을 저장하세요:
# Hello Python / I am AI Software

with open("output/helloEx.txt", "w", encoding="utf-8") as a:
    a.write("Hello Python\n")
    a.write("I am AI Software\n")
```

```python
# 과제 2번.
# 학생 이름 5명을 입력받아 student.txt에 저장하세요.

students = ["이0우", "홍0동", "류0희", "김0훈", "최0환"]
with open("output/studentsEx.txt", "w", encoding="utf-8") as b:
    for i in students:
        b.write(i+"\n")
```

```python
# 과제 3번.
# student.txt의 내용을 읽어 화면에 모두 출력하세요.

with open("output/studentsEx.txt", "r", encoding="utf-8") as c:
    for i in c:
        print(i)
```

```python
# 과제 4번.
# 사용자의 이름과 나이를 입력받아 member.txt에 홍길동,20 형태로 저장하세요.

name = input("이름: ")
age = input("나이: ")
with open("output/member.txt", "w", encoding="utf-8") as d:
    d.write(name+"\n")
    d.write(age+"\n")
```

```python
# 과제 5번.
# fruit.txt(사과, 배, 포도, 복숭아)를 생성하고 반복문으로 출력하세요.

fruit = ["사과","배","포도","복숭아"]
with open("output/fruit.txt","w+",encoding="utf-8") as f:
    for i in fruit:
        f.write(i+", ")
    f.seek(0)
    print(f.read())
```

```python
# 과제 6번.
# 학생 점수 5개를 입력받아 파일에 저장한 후, 평균을 계산하여 출력하세요.

students6 = []
for i in range(5):
    inp = int(input(f"{i+1}번째 학생 점수: "))
    students6.append(inp)
print(students6)
with open("output/students6.txt","w",encoding="utf-8") as g:
    for i in students6:
        g.write(str(i) + ", ")
average = sum(students6) / len(students6)
print("평균: "+average)
```

```python
# 과제 7.
# CSV 파일(이름,Python,AI,Web 헤더 포함)을 생성하세요.

with open("output/csv6.csv","w",encoding="utf-8") as h:
    h.write("이름")
    h.write("Python")
    h.write("AI")
    h.write("Web")
```

```python
# 과제 8.
# source.txt의 내용을 backup.txt로 복사하는 프로그램을 작성하세요.

with open("output/source.txt", "r", encoding="utf-8") as source:
    with open("output/backup.txt", "w", encoding="utf-8") as backup:
        backup.write(source.read())
```

```python
# 과제 9.
# 존재하지 않는 파일을 읽을 때 예외 처리를 적용하여 "파일이 존재하지 않습니다."를 출력하세요.

try:
    with open("output/test.txt", "r", encoding="utf-8") as i:
        print(i.read())
except FileNotFoundError:
    print("파일이 존재하지 않습니다.")
```

```python
# 과제 10.
# 학생 정보를 입력받아 저장하고, 저장된 내용을 다시 읽어 아래와 같은 표 형태로 출력하세요:
# ========================================
# 이름        점수
# ========================================
# 홍길동      95
# 김철수      88
# 이영희      100
# ========================================

names = []
scores = []
for i in range(1,4):
    name2 = input(str(i) + "번째 이름: ")
    score2 = int(input(str(i) + "번째 점수: "))

    names.append(name2)
    scores.append(score2)

print("========================================")
print("이름\t\t점수")
print("========================================")
for i, j in zip(names, scores):
    print(f"{i}\t\t{str(j)}")
print("========================================")
```
