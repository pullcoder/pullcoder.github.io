```python
# 과제 3
# 학생 정보 딕셔너리
# 학생 정보를 딕셔너리로 저장하세요.

# 이름
# 학번
# 학과
# 학년
# 평균평점

dick = {"이름":"이삼사", "학번":"20260000", "학과":"자율전공", "학년":0, "평균평점":0.0}
for k, v in dick.items():
    print(f"{k}\t{v}")

dick["이름"] = "이정우"
dick["학번"] = "20260790"
dick["학과"] = "AI소프트웨어"
dick["학년"] = 1
dick["평균평점"] = 4.0
print("-----------------")
for k, v in dick.items():
    print(f"{k}\t{v}")
```

```python
# 과제 4
# 리스트 평균
# score = [95,88,100,75,82]
# 위 리스트의 평균을 구하세요.

score = [95,88,100,75,82]
print(sum(score)/len(score))
```

```python
# 과제 5
# 리스트 반복 출력
# fruit = ["사과","배","포도","복숭아"]
# 반복문으로 출력하세요.

fruit = ["사과","배","포도","복숭아"]

for i in fruit:
    print(i, end=" ")
```

```python
# 과제 6
# Key / Value 출력
# student = {"이름":"홍길동","학과":"AI","학년":1}
# 모든 Key와 Value를 출력하세요.

student = {"이름":"홍길동","학과":"AI","학년":1}

for k, v in student.items():
    print(f"{k}\t{v}")
```

```python
# 과제 7
# 리스트 안의 딕셔너리 저장
# 이름	Python	AI
# 홍길동	95	90
# 김철수	88	85
# 이영희	100	98

column = ["이름", "Python", "AI"]

dick1 = {column[0]: "홍길동", column[1]: 95, column[2]: 90}
dick2 = {column[0]: "김철수", column[1]: 88, column[2]: 85}
dick3 = {column[0]: "이영희", column[1]: 100, column[2]: 98}

dick = [dick1, dick2, dick3]

print(dick)

print("---------------")
for i in column:
    print(i,end="\t")
print()
for i in dick:
    for k, v in i.items():
        print(v, end="\t")
    print()
```

```python
# 과제 8
# 총점 / 평균 / 최고점 / 최저점
# 학생 5명의 점수를 리스트에 저장한 후 총점, 평균, 최고점, 최저점을 출력하세요.

students = [97, 82, 90, 67, 76]

print(f"총점\t{sum(students)}")
print(f"평균\t{sum(students)/len(students)}")
print(f"최고\t{max(students)}")
print(f"최저\t{min(students)}")
```

```python
# 과제 10
# 평균 점수 계산
# students = [
#   {"이름":"홍길동","점수":95},
#   {"이름":"김철수","점수":88},
#   {"이름":"이영희","점수":100}
# ]
# 학생 3명의 정보를 저장하고 평균 점수를 계산하세요.

students10 = [
  {"이름":"홍길동","점수":95},
  {"이름":"김철수","점수":88},
  {"이름":"이영희","점수":100}
]

score10 = []

for i in students10:
    score10.append(i["점수"])

print(score10)

avg10 = sum(score10) / len(score10)
print("평균 점수:", avg10)
```
