```python
s = "abc"
s.upper()          # 새 문자열 "ABC"를 만들 뿐, s 자신은 그대로 "abc"
print(s)            # abc

l = [3, 1, 2]
l.sort()            # 원본 l 자체를 정렬함
print(l)            # [1, 2, 3]
```

```python
title = "python programming"
print(title.title())        # Python Programming

score = "7"
print(score.zfill(3))       # 007

raw = "   note   "
print(raw.strip())          # "note"

text = "  hello Python  "

print(text.upper())       # "  HELLO PYTHON  "
print(text.lower())       # "  hello python  "
print(text.capitalize())  # "  hello python  "
print(text.title())       # "  Hello Python  "
print("PyThOn".swapcase())  # "pYtHoN"

print(text.strip())       # "hello Python"
print("...hello...".strip("."))  # "hello"
print("  hello  ".lstrip())      # "hello  "
print("  hello  ".rstrip())      # "  hello"

print("42".zfill(5))     # "00042"
print("-42".zfill(5))    # "-0042" 
```

```python
def get_age(text):
    if not text.isdigit():
        raise ValueError("숫자만 입력하세요")
    return int(text)

print(get_age("24"))     # 24
```

```python
members = [("지훈", 22), ("서연", 20), ("민재", 25)]

members.sort(key=lambda m: m[1])          # 나이 기준 오름차순
print(members)   # [('서연', 20), ('지훈', 22), ('민재', 25)]

members.sort(key=lambda m: m[1], reverse=True)  # 내림차순
```

```python
grades = [90, 85, 90, 70]
print(grades.index(90))   # 0
print(grades.count(90))   # 2

backup = grades.copy()    # 원본과 독립된 새 리스트
backup.append(100)
print(grades, backup)     # [90, 85, 90, 70] [90, 85, 90, 70, 100]
```

```python
python_class = {"지훈", "서연", "민재"}
ai_class = {"서연", "하은"}

print(python_class & ai_class)     # {'서연'}  두 수업 모두 듣는 학생
print(python_class | ai_class)     # 합집합
print(python_class - ai_class)     # 파이썬 수업만 듣는 학생
```

```python
def check_password(pw):
    if len(pw) < 8:
        return "약함: 8자 이상이어야 합니다."
    if pw.isdigit() | pw.isalpha():
        return "약함: 숫자와 문자를 섞으세요."
    if pw.islower() | pw.isupper():
        return "보통: 대소문자를 섞으면 더 안전합니다."
    return "강함."

password = ["abc123", "12345678", "abcd1234", "ABCd1234"]

for pw in password:
    print(pw, "->", check_password(pw))
```
